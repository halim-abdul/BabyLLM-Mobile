# Incremental memory baseline
Toy demonstration only. Train and evaluation text overlap here, so the score is **not** a held-out result.

```python
from babyllm import NGramMemory
from babyllm.evaluation import evaluate_next_token

memory = NGramMemory(order=3)
memory.learn('hello world', document_id='example-1')
memory.learn('hello researcher', document_id='example-2')
print(memory.statistics())
print(memory.candidates(['<s>', 'hello']))
print(memory.complete('hello'))
```

```python
# Replace with a genuinely held-out set before reporting research results.
print(evaluate_next_token(memory, ['hello world']))
memory.close()
```
